## AT Task 1: Add a Method That Uses a Condition

```python
class Animal:
    def __init__(self, name, species, age, adopted = False):
        self.name = name
        self.species = species
        self.age = age
        self.adopted = adopted
    def describe(self):
        return f"{self.name} is a {self.age}-year-old {self.species}"
    def adopt(self):
        self.adopted = True
        return(f"{self.name} has been adopted!")
    def status(self):
        if self.adopted is True:
            return f"{self.name} has found a home."
        else:
            return f"{self.name} is still waiting."

shelter = [Animal("Squeakers", "Cat", 1), Animal("Fuzzy", "Dog", 1), Animal("Steve", "Cat", 2), Animal("Mike", "Cat", 3)]
for Animal in shelter:
    print(Animal.describe())
    print(Animal.status())

```

## AT Task 2: Count Unadopted Animals

```python

available_count = 0

for Animal in shelter:
    if Animal.adopted == False:
        available_count = available_count + 1

print(f"Animals still available: {available_count}")
```

## Extra Credit Task: Add a notes Attribute and an add_note method

```python
class Animal:
    def __init__(self, name, species, age, adopted = False):
        self.notes = []
        self.name = name
        self.species = species
        self.age = age
        self.adopted = adopted
    def describe(self):
        return f"{self.name} is a {self.age}-year-old {self.species}"
    def adopt(self):
        self.adopted = True
        return(f"{self.name} has been adopted!")
    def status(self):
        if self.adopted is True:
            return f"{self.name} has found a home."
        else:
            return f"{self.name} is still waiting."
    def add_note(self, text):
        self.notes.append(text)
        print("note added")
    def show_notes(self):
        if len(self.notes) == 0:
            print("No notes on file.")
        else:
            for notes in self.notes: 
                print("-", notes)

animal_1.add_note("Squeakers squeaks")
animal_2.add_note("Fuzzy barks")
animal_3.add_note("Steve meows")

animal_2.show_notes()

```
