### AT 1

```python
class Animal:
    def __init__(self, name, species, age, adopted = False):
        self.name = name
        self.species = species
        self.age = age
        self.adopted = adopted

    def describe(self):
        return f"{self.name} is a {self.age}-year-old {self.species}, has been adopted: {self.adopted}."
    
    def adopt(self):
        self.adopted = True
        return (f"{self.name} has been adopted!")
    
    def status(self):
        if self.adopted == True:
            return f"{self.name} has found a home!"
        else:
            return f"{self.name} is still waiting..."

for animal in shelter:
    print(animal.status())
```

```python
# A function that calculates a final exam score
# by dropping the lowest quiz score first
def final_score(quizzes, exam):
    print("quizzes received:", quizzes)   # check input
    print("exam received:", exam)         # check input

    # Remove the lowest quiz score
    quizzes.remove(min(quizzes))
    print("quizzes after drop:", quizzes) # check after change

    # Average the remaining quizzes
    quiz_avg = sum(quizzes) / len(quizzes)
    print("quiz average:", quiz_avg)      # check intermediate

    # Final score: 60% exam, 40% quiz average
    score = (exam * 0.6) + (quiz_avg * 0.4)
    print("final score:", score)          # check result
    return score

quiz_grades = [70, 85, 90, 60]
exam_grade = 88

result = final_score(quiz_grades, exam_grade)
print("Returned value:", result)
```
