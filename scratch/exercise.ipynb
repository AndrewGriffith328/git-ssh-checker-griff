```python
# A function with two int parameters
def calculate_area(length, width):
    return length * width

# A function with two float parameters
def calculate_average(score1, score2):
    return (score1 + score2) / 2

# A function with a str and an int parameter
def make_profile(name, age):
    return name + " is " + str(age) + " years old."

# A function with a str and a bool parameter
def display_status(username, is_active):
    if is_active:
        return username + " is currently active."
    return username + " is inactive."

# Calling each function with arguments in the right order
area = calculate_area(10, 4)
print(area)                             # prints: 40

avg = calculate_average(88.5, 91.0)
print(avg)                              # prints: 89.75

profile = make_profile("Taylor", 37)
print(profile)      # prints: Taylor is 37 years old.

status = display_status("alex99", False)
print(status)       # prints: alex99 is currently active.
```

```python
# Define double_score
def double_score(score):
    return score * 2

# Define cap_score
def cap_score(score):
    return min(score, 100)

# Define label_score
def label_score(score):
    return "Score: " + str (score)

# Compose all three for a starting score of 40
# and print the result

print(label_score(cap_score(double_score(40))))
```

```python
def process_score(score):
    doubled = score * 2
    capped = min(doubled, 100)
    return "Score: " + str(capped)

print(process_score(40))
```
